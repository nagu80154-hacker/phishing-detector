```python
def analyze_url(url):

    if not url.startswith(("http://", "https://")):
        url = "https://" + url

    parsed = urlparse(url)
    host = parsed.netloc.lower()

    score = 0
    reasons = []

    if len(url) > 100:
        score += 15
        reasons.append("URL is unusually long")

    if "@" in url:
        score += 20
        reasons.append("URL contains @ symbol")

    if is_shortened_url(host):
        score += 20
        reasons.append("URL uses a shortening service")

    if is_ip_address(host):
        score += 25
        reasons.append("URL uses an IP address instead of a domain")

    if host.count(".") >= 3:
        score += 10
        reasons.append("URL has many subdomains")

    if host.count("-") >= 2:
        score += 10
        reasons.append("Domain contains multiple hyphens")

    suspicious_words = [
        "login",
        "verify",
        "secure",
        "account",
        "update",
        "password",
        "banking"
    ]

    for word in suspicious_words:
        if word in url.lower():
            score += 10
            reasons.append(f"Contains suspicious keyword: {word}")
            break

    if not url.startswith("https://"):
        score += 10
        reasons.append("URL does not use HTTPS")

    score = min(score, 100)

    if score >= 50:
        result = "🚨 Potentially Phishing"
    elif score >= 25:
        result = "⚠️ Needs Caution"
    else:
        result = "✅ Likely Legitimate"

    return url, host, score, result, reasons
```
